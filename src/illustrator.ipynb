# Sensor stream

This notebook connects to the `sensor` container and updates a Plotly figure with live temperature data.

```python
import socket
import time
from collections import deque

import plotly.graph_objects as go
from IPython.display import display

SENSOR_HOST = "sensor"
SENSOR_PORT = 1234
MAX_POINTS = 200
REFRESH_SECONDS = 0.1
```

```python
x_values = deque(maxlen=MAX_POINTS)
temperatures = deque(maxlen=MAX_POINTS)

fig = go.FigureWidget()
fig.add_scatter(x=[], y=[], mode="lines+markers", name="Temperature")
fig.update_layout(
    title="Live temperature stream",
    xaxis_title="Sample",
    yaxis_title="Temperature (C)",
    template="plotly_white",
)

display(fig)
```

```python
sample = 0

with socket.create_connection((SENSOR_HOST, SENSOR_PORT), timeout=10) as sock:
    reader = sock.makefile("r", encoding="utf-8")

    while True:
        line = reader.readline()
        if line == "":
            raise ConnectionError("Sensor connection closed")

        sample += 1
        x_values.append(sample)
        temperatures.append(float(line.strip()))

        with fig.batch_update():
            fig.data[0].x = list(x_values)
            fig.data[0].y = list(temperatures)

        time.sleep(REFRESH_SECONDS)
```
